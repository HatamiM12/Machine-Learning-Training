```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Generate Synthetic Data
def generate_data(n_samples=75, noise=10000.0):
  np.random.seed(42)
  X = np.linspace(0, 50, n_samples)
  # Function = 3x^3 + x^2 - 5x + 2
  true_slope1 = 3
  true_slope2 = 1
  true_slope3 = -5
  true_intercept = 2
  noise = np.random.randn(n_samples) * noise
  y = (true_slope1 * X ** 3) + (true_slope2 * X ** 2) + (true_slope3 * X)+ true_intercept + noise
  return X, y

X, y = generate_data()
```

```python
#Function to generate polynomial features
def polynumial_features(X, degree):
  X_ploy = np.c_[np.ones(len(X))]
  for i in range(1, degree + 1):
    X_poly = np.c_[X_ploy, X**i]
  return X_poly
```

```python
#Define polynomial function
def polynomial_regression(X, y, degree):
  X_poly = polynumial_features(X, degree) #Consider X_poly as X'
  #W = (X'^T * X')^-1 * X'^T * y , using pseudoinverse in this fomula
  w = np.linalg.pinv(X_poly).dot(y)
  return w
m = 3
w_poly = polynomial_regression(X, y, m)
print(f"Parameters (w) for degree {m}: {w_poly}")
```

```python
#Visualizing polynomial
X_fit = np.linspace(X.min(), X.max(), 200)
X_fit_poly = polynumial_features(X_fit, m)
y_poly_pred = X_fit_poly.dot(w_poly)
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X_fit, y_poly_pred, color='green', label=f'Polynomial Fit (Degree {m})')
plt.title(f"Polynomial Regression (Degree {m})")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.legend()
plt.show()
```
